```python
import pandas as pd 
import matplotlib.pyplot as plt 
import sqlite3
import seaborn as sns
```

```python
import os

data_path = "../Data"
figure_path = "../figures"
database_path = "../Database"
os.makedirs(figure_path , exist_ok = True) # make directory 

## create database path & connector
db_path = f"{database_path}/esg_database_reserved.db"
conn = sqlite3.connect("../Database/esg_database_reserved.db")

## write run_sql command to use sql command
def run_sql(query_cmd , connector) : 
    """Run a SQL query and return the result as a DataFrame."""
    return pd.read_sql_query(sql = query_cmd , con = connector)

def read_and_load_csv_to_db(file_path , connector , table_name , encoding) : 
    # read csv (with encoding)
    df = pd.read_csv(file_path , encoding = encoding)
    # load csv to db
    df.to_sql(name = table_name , con = connector , if_exists = "replace" , index = False) 
    
def execute_sql(command , connector = conn) : 
    connector.execute(command) 
    connector.commit() # to change data 
    
## run_sql wouldn't change data , while execute_sql would change data forever
```

```python
# build super database (reserved)
raw_files = {
    "company_info":     ("../Data/company_info.csv", "utf-8-sig"),
    "company_perf" :    ("../Data/company_perf.csv", "utf-8-sig")
}

for each_file in raw_files : 
    read_and_load_csv_to_db(file_path = raw_files[each_file][0] , connector = conn , 
                            table_name = each_file , encoding = raw_files[each_file][1] )
    
# show the tables in the database
run_sql("SELECT name FROM sqlite_master WHERE type='table'" , connector = conn)
```

## WHERE vs. HAVING 
- WHERE → Filters individual rows.

- HAVING → Filters groups after aggregation

```python
## Question → What is the average ESG score for each company? 
run_sql("SELECT company_id , AVG(esg_overall) AS avg_esg_overall FROM company_perf GROUP BY company_id HAVING AVG(esg_overall) > 70" ,
        connector = esg_conn) ## gb then ob
```

## Why Do We Need JOIN?

➔ INNER JOIN : Matching rows from both tables

➔ LEFT JOIN : All rows from left + matching rows from right

➔ RIGHT JOIN : All rows from right + matching rows from left

➔ FULL OUTER JOIN : All rows from both tables

### INNER JOIN (Default of JOIN)

```python
inner_join_cmd = """
SELECT
    c.company_name,
    p.year,
    p.revenue,
    p.esg_overall
FROM company_info AS c
INNER JOIN company_perf AS p
    ON c.company_id = p.company_id
"""

run_sql(inner_join_cmd , connector = esg_conn)
```

## JOIN + WHERE 

```python
## Question → Which companies had the highest revenue in 2025? 
run_sql("""
SELECT 
    c.company_name , p.year , p.revenue 
FROM company_info AS c 
JOIN company_perf AS p 
    ON c.company_id = p.company_id 
WHERE p.year = 2025         
ORDER BY p.revenue DESC 
LIMIT 10
""" , connector = esg_conn) 
```

## JOIN + GROUP BY 

```python
## Question → What is the average ESG score for each industry?
run_sql("""
SELECT c.industry , AVG(p.esg_overall) as average_esg
FROM company_info AS c
JOIN company_perf as p 
    ON c.company_id = p.company_id
GROUP BY c.industry
ORDER BY average_esg DESC 
""" , connector = esg_conn) 
```

## JOIN + GROUP BY + HAVING 

```python
## Question → Which industries have an average ESG score greater than 60? 

run_sql("""
SELECT c.industry , AVG(p.esg_overall) as average_esg
FROM company_info AS c
JOIN company_perf as p 
    ON c.company_id = p.company_id
GROUP BY c.industry
HAVING AVG(p.esg_overall) > 60
ORDER BY average_esg DESC 
""" , connector = esg_conn) 
```

## SQL Command Categories 


Category | Full name | Purpose | Examples
| :--- | :---: | :---: | ---: |
DDL | Data Definition Language | Define database structure | CREATE, ALTER, DROP
DML | Data Manipulation Language | Modify data | INSERT, UPDATE, DELETE
DQL | Data Query Language | Retrieve data | SELECT
DCL | Data Control Language | Control access/permissions | GRANT, REVOKE
TCL | Transaction Control Language | Manage transactions | COMMIT, ROLLBACK

## DDL — Data Definition Language 
DDL is used to define and modify the structure of a database.

### CREATE (No code)

### ALTER
ALTER is used to modify an existing database structure. 

```python
run_sql("""
ALTER TABLE company_info
ADD COLUMN founded_year INTEGER;
""" , connector = esg_conn) 
```

### DROP
DROP removes a database object. 

```python
run_sql("""
DROP TABLE company_info;
""" , connector = esg_conn) 
```

## DML — Data Manipulation Language
DML is used to add, modify, and remove data in tables.
